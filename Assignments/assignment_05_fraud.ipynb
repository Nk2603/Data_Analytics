## Credit Card Fraud Detection


<img src="./assignment_05_ss/Screenshot 2026-03-09 114803.png"/>

**Explanation**
* **Dashboard Overview:** The dashboard analyzes **credit card transactions to identify fraudulent activities**, providing a quick overview of transaction counts and fraud detection insights.

* **Fraud vs Non-Fraud Chart:** The column chart displays the **count of transactions based on the fraud label (IsFraud)**, where `0` represents normal transactions and `1` represents fraudulent transactions.

* **Key Metrics Cards:** The cards highlight **Total Transactions (100K)** and **Total Fraudulent Transactions (1000)**, helping users quickly understand the scale of fraud in the dataset.

* **Transaction Details Table:** The table lists **transaction amount, transaction date, and fraud status**, allowing users to examine individual transactions and identify suspicious ones.

* **Fraud Filter (Slicer):** The **IsFraud filter** allows users to interactively view either **fraudulent or non-fraudulent transactions**, making the analysis more focused and flexible.
