<img src="./assignment_04_screenshots/Screenshot 2026-02-24 115550.png"/>

**Aggregate Vs Scalar function :**
- An aggregate function operates on multiple rows at the same time and returns a single result. A scalar   function operates on one row at a time and returns one result for each row.

- Generally, aggregate functions are used on numeric values (for example, SUM(), AVG(), COUNT()), while scalar functions are commonly used on character data such as VARCHAR to perform operations like UPPER().

In SQL, NULL values are not actual values — they represent unknown or missing data in a column.

You cannot check NULL using operators like = or !=.

The correct way to check for NULL is:

-  To find NULL values
   - WHERE column_name IS NULL;

-  To find NOT NULL values
   - WHERE column_name IS NOT NULL;

**Question-10 :** Write a query to find employees who earn more than the average 
salary of their department (without using joins). 
Explain the logic of the subquery

**Query : SELECT * FROM employees e WHERE salary > (SELECT AVG(salary) FROM employees WHERE department = e.department);**

<img src="./assignment_04_screenshots/Screenshot 2026-02-24 181459.png"/>

logic of subquery : 
- A subquery is a query inside another query.

- In this query, it’s a correlated subquery because it depends on the outer query (e.department).

- The subquery calculates the average salary of the current employee’s department.

- The outer query compares each employee’s salary to this average.

- Only employees earning more than their department’s average are returned.
